```python
from pdf2image import convert_from_bytes
import pytesseract
import os
from pathlib import Path
from pytesseract import Output
import pandas as pd

from config import POPPLER_PATH, TESSERACT_CMD, TESSDATA_PREFIX
os.environ["TESSDATA_PREFIX"] = TESSDATA_PREFIX
pytesseract.pytesseract.tesseract_cmd = TESSERACT_CMD
```

```python
def pdf_to_text(pdf_bytes):
    pages = convert_from_bytes(
        pdf_bytes,
        dpi=300,
        poppler_path=POPPLER_PATH
    )

    full_text = "\n\n".join(
        pytesseract.image_to_string(page, lang="rus")
        for page in pages
    )

    return full_text
```

```python
pdf_path = Path("doci_primery") / "1. ПрофиМаркет_30.03.2026.pdf"
with open(pdf_path, "rb") as f:
    pdf_bytes = f.read()

text = pdf_to_text(pdf_bytes)
```

```python
from pytesseract import Output
import pandas as pd
pages = convert_from_bytes(
	pdf_bytes,
	dpi=300,
	poppler_path=POPPLER_PATH
)
data = pytesseract.image_to_data(
    pages[0],
    lang="rus+eng",
    config="--oem 3 --psm 6",
    output_type=Output.DATAFRAME
)

data = data.dropna(subset=["text"])
data = data[data["conf"] > 30]

data[["text", "left", "top", "width", "height", "conf"]].head()
```

```python
from pathlib import Path
from img2table.document import PDF
from img2table.ocr import TesseractOCR

from config import TESSDATA_PREFIX

pdf_path = Path("doci_primery") / "1. ПрофиМаркет_30.03.2026.pdf"
out_path = Path("doci_primery") / "profimarket_tables.xlsx"

ocr = TesseractOCR(
    n_threads=1,
    lang="rus+eng",
    tessdata_dir=TESSDATA_PREFIX
)

doc = PDF(
    src=str(pdf_path),
    pdf_text_extraction=False
)

doc.to_xlsx(
    dest=str(out_path),
    ocr=ocr,
    implicit_rows=True,
    borderless_tables=True,
    min_confidence=40
)
```
